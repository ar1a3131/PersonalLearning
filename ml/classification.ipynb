# **Notes on IBM Classification Course**

### **Types of Supervised Learning**

- **Regression:** outcome is continuous/numerical (e.g.: house pricing, box office revenues, event attendance, netwwork load, portfolio losses)
- **Classification:** outcome is a category (e.g.: detect fraud, customer churn, event atendance or network load [whether they pass a certain threshold], loan default)


## **What is needed for classification?**

- Data Cleaning (missing value imputation, outlier removal, duplicate removal)

- Data Transformation & Scaling (feature scaling [StandardScaler or MinMaxScaler], categorical encoding [One-Hot Encoding], Binning [group continuous numbers into discrete intervals to reduce noise/simplify complex patterns])

- Data Reduction & Balancing (dimensionality reduction [PCA], class balancing [SMOTE], data splitting)

- allow learning from known examples (training set (70-80% of data to teach the model) + testing set (20-30% to evaluate it))



## **List of Supervised Learning Model**
### each of these models can be used for both regression and classification

1. logistic regression: predicts a discrete categorical class (usually binary) (while linear regression predicts a continuous numeric value)

2. k-nearest neighbors: nonlinear and simplistic approach to categorizing according to the similarity of past examples nearest to the feature space of the label we're trying to predict

3. support vector machines: powerful linear classifier, which will leverage the **kernal trick** (shortcut used in ML to help linear algorithms solve non-linear problems w/out causing computer resource exhaustion) to allow for complex decision boundaries

4. neural networks
5. decision tree
6. random forests
7. boosting
8. ensemble models

#### **the kernel trick (if you're curious)**

The Kernel Trick is just a formula (the specific formula depends on the situation):
- Linear Kernel (No trick) [straight forward dot product... simple, straight-line separation]: $$K(x_1, x_2) = (x_1 \cdot x_2)$$

- Polynomial Kernel (A single equation that calculates curved, polynomial relationships (where d is the degree, like squaring or cubing the data)): $$K(x_1, x_2) = (x_1 \cdot x_2 + c)^d$$

- RBF (radial basis function)/Gaussian Kernel (A single equation based on exponentiation. This is the formula that allows the model to act as if it is working in an infinite-dimensional space, creating tight circular boundaries around data points): $$K(x_1, x_2) = \exp(-\gamma ||x_1 - x_2||^2)$$



exp = the exponential function ($e^z$) where e is Euler's number, it acts as a **similarity translator**, the value inside the exponent is always negative... when points are identical ($x_1 = x_2$), the formula becomes exp(0), which is exactly 1 (maximum similarity)... as the points move further apart, the exponent gets more negative and therefore closer to 0

γ = radius of influence, you can have a low γ (less sensitive, large and wide radius of influence) or a high γ (more sensitive, tiny and sharp radius of influence)... if the γ is too high, you can get severe overfitting...
