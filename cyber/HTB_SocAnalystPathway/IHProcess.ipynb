# **Incident Handling (IH)**
### SOC Analyst Pathway (Hack the Box), Section 1




**Event:** an action occurring in a system or network, e.g.: a user sending an email, a mouse click, a firewall allowing a connection request

**Incident:** an event with negative consequences, e.g.: a system crash, unauthorized access to sensitive data... can also occur due to natural disasters, power failures, etc.

**IT Security Incident:** an event with clear intent to cause harm, e.g.: data theft, funds theft, unauthorized access to data, installation and use of malware and remote access tools



####  A comprehensive **incident handling plan** should address various types of incidents and provide appropriate measures to identify, contain, eradicate, and recover

One of the most widely used resource for incident handling: https://nvlpubs.nist.gov/nistpubs/SpecialPublications/NIST.SP.800-61r3.pdf

## Different Types of Real-World Incidents

- Leaked Credentials; e.g.: the Colonial Pipeline (major American oil pipeline system) Ransomware Attack (2021), originated from a leaked password for an inactive VPN (Virtual Private Network) account, which did not have Multi-Factor Authentication (MFA) enabled
(https://en.wikipedia.org/wiki/Colonial_Pipeline_ransomware_attack)

- Default / Weak Credentials; e.g.: Mirai Botnet (2016) scanned for IoT devices using factory credentuals (admin/admin) and turned them into a massive DDoS botnet. **[DDos: Distrubuted DoS - attack comes from many sources, Dos: Denial of Service - comes from one source]**
