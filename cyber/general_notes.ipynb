# types of encryption

- Symmetric: same key used for encryption and decryption
- Assymetric: uses a public key and a private key
- AES (Advanced Encryption Standard): fast and widely used for securing data
- RSA (Rivest Shamir Adleman): public-key encryption commonly usde for secure communication
- ECC (Elliptic-Curve Cryptography): strong public-key cryptography with smaller keys
- CHACHA20: fast modern encryption desgined for secure data transfer
- 3DES (Triple Data Encryption Standard): older encryption standaard; now considered legacy
- Blowfish: a symmetric cipher designed for fast encryption
- Twofish: A strong encryption cipher and AES finalist
- One-Time Pad: uses a truly random key only once for encryption
